```python
import sympy
```

```python
J, w = sympy.symbols('J ,w')
```

```python
J  = w**3
```

```python
J
```

```python
dj_dw = sympy.diff(J,w)
dj_dw
```

```python
dj_dw.subs([(w,2)])
```

```python
dj_dw2 = sympy.diff(dj_dw,w)
```
